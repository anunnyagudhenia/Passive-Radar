```python
import numpy as np
import math
import matplotlib.pyplot as plt
```

```python
def Signal_to_noise_ratio(Pt,Gt,Gr,sigma,wavelength,N,D,tau,Rt,Rr,B):
    wav=wavelength**2
    pi=(4*np.pi)**3
    Constant = k*Temp
    snr=(Pt*Gt*Gr*sigma*wav*N*D*np.sqrt(tau))/(pi*math.pow(Rt,2)*math.pow(Rr,2)*Constant*np.sqrt(B))
    return snr
```

```python
Pt = 235*1e3       #Transmitted Power of Aries
Gt = 2511.9         #Linear gain of Aries 34dbi
f = 206.5 *1e6     #Frequency of Aries
c=3*1e8
wavelength = c/f
Rt = 3*1e6         #Distance between target and transmitter
Rr = 3*1e6         #Distance between target and receiver
sigma  = 1          #RCS
D = 0.13            #Duty Cycle of Aries
k = 1.38*1e-23
Temp = 290
```

```python
#Received Power for isotropic antenna
Gr=1
def received_power(Pt,wavelength,Gt,Gr,Rt,Rr):
    Pr = (Pt*Gt*math.pow(wavelength,2)*Gr*sigma)/(math.pow(4*np.pi,3)*math.pow(Rt,2)*math.pow(Rr,2))
    return Pr


Pri = received_power(Pt,wavelength,Gt,Gr,Rt,Rr)
print("Recived Power for Isotropic receiver antenna = ",Pri)
```

```python
# Received Power for Directive Antenna
Gr = 12.6
Prd = received_power(Pt,wavelength,Gt,Gr,Rt,Rr)
print("Recived Power for Directive receiver antenna = ",Prd)
```

```python
#SNR vs RCS
N = 10
tau = 0.5
B = 2*1e4
rcs=np.linspace(1,1000,10000)
snr=[]
for i in rcs:
    snr.append(Signal_to_noise_ratio(Pt,Gt,Gr,i,wavelength,N,D,tau,Rt,Rr,B))

plt.plot(rcs,snr)
plt.grid()
plt.xscale('log')
plt.yscale('log')
plt.xlabel("RCS")
plt.ylabel("SNR")
```

```python
def num_antenna(SNR,Pt,Gt,Gr,sigma,wavelength,D,tau,Rt,Rr,B):
    wav=wavelength**2
    pi=(4*np.pi)**3
    Constant = k*Temp
    N = (SNR*pi*math.pow(Rt,2)*math.pow(Rr,2)*Constant*math.pow(B,0.5))/(Pt*Gt*Gr*sigma*wav*D*math.pow(tau,0.5))
    return N
```

```python
#SNR vs Bandwidth
Pt = 235*1e3       #Transmitted Power of Aries
Gt = 2511.9         #Linear gain of Aries 34dbi
f = 206.5 *1e6     #Frequency of Aries
c=3*1e8
wavelength = c/f
Rt = 3*1e6         #Distance between target and transmitter
Rr = 3*1e6         #Distance between target and receiver
sigma  = 1          #RCS
D = 0.13            #Duty Cycle of Aries
k = 1.38*math.pow(10,-23)
Temp = 290
samples = 1e6
SNR = 10
B = np.linspace(16000,2*1e6)
N = []
for i in B:
    tau = samples/i
    N.append(num_antenna(SNR,Pt,Gt,Gr,sigma,wavelength,D,tau,Rt,Rr,i))

plt.plot(B,N)
plt.grid()
plt.xlabel("B")
plt.ylabel("N")
```

```python
# -----------------------------
# N vs Bandwidth for different samples
# -----------------------------
samples_list = [1e5, 1e6, 1e7, 1e8]
B = np.linspace(1.6e4, 2e6, 300)

plt.figure()

for samples in samples_list:
    N_values = []
    
    for b in B:
        tau = samples / b   # n = B * tau
        N_val = num_antenna(SNR, Pt, Gt, Gr, sigma,
                            wavelength, D, tau, Rt, Rr, b)
        N_values.append(N_val)
    
    plt.plot(B, N_values, label=f"n = {samples:.0e}")

plt.grid()
plt.xlabel("Bandwidth (Hz)")
plt.ylabel("Number of Antennas (N)")
plt.legend()
plt.show()
```

```python
#minimum detectable RCS
def min_RCS(SNR,Pt,Gt,Gr,wavelength,D,tau,Rt,Rr,B,N):
    wav=wavelength**2
    pi=(4*np.pi)**3
    Constant = k*Temp
    min_sigma = (SNR*pi*math.pow(Rt,2)*math.pow(Rr,2)*Constant*math.pow(B,0.5))/(Pt*Gt*Gr*wav*D*math.pow(tau,0.5)*N)
    return min_sigma

Pt = 235*1e3       #Transmitted Power of Aries
Gt = 2511.9         #Linear gain of Aries 34dbi
f = 206.5 *1e6     #Frequency of Aries
c=3*1e8
wavelength = c/f
Rt = 3*1e6         #Distance between target and transmitter
Rr = 3*1e6         #Distance between target and receiver
D = 0.13            #Duty Cycle of Aries
k = 1.38*math.pow(10,-23)
Temp = 290
tau = 0.5
B = 2*1e7
N=np.linspace(2,100,99)
rcs=[]
for i in N:
    rcs.append(min_RCS(SNR,Pt,Gt,Gr,wavelength,D,tau,Rt,Rr,B,i))

plt.plot(N,rcs)
plt.grid()
plt.xscale('log')
plt.yscale('log')
plt.xlabel("Number of Antennas")
plt.ylabel("minimum detetctable rcs")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import math

# ==============================
# SYSTEM PARAMETERS
# ==============================
Pt = 235e3                 # Transmit Power
Gt = 2511.9
Gr = Gt
f = 206.5e6
c = 3e8
wavelength = c / f

Rt = 3e6
Rr = 3e6
D = 0.13
tau = 1
B = 1e6
SNR_required = 10

k = 1.38e-23
Temp = 290

# ==============================
# FUNCTION: NUMBER OF ANTENNAS
# ==============================
def num_antenna(SNR, Pt, Gt, Gr, sigma, wavelength, D, tau, Rt, Rr, B):
    
    # Received power (bistatic radar equation)
    Pr = (Pt * Gt * Gr * (wavelength**2) * sigma) / \
         ((4 * np.pi)**3 * (Rt**2) * (Rr**2))
    
    # Noise power
    Noise = k * Temp * B
    
    # Single antenna SNR
    snr_single = Pr / Noise
    
    # Coherent integration gain (N antennas)
    # SNR_total = N * snr_single * D * tau
    N = SNR / (snr_single * D * tau)
    
    return N


# ==============================
# RCS RANGE FOR CURVE
# ==============================
sigma_vals = np.logspace(-6, 3, 600)   # very small to large RCS

N_vals = []
for s in sigma_vals:
    N_vals.append(
        num_antenna(SNR_required, Pt, Gt, Gr, s,
                    wavelength, D, tau, Rt, Rr, B)
    )

N_vals = np.array(N_vals)


# ==============================
# LEO SIZE DISTRIBUTION
# ==============================
np.random.seed(0)

small = np.random.uniform(1e-3, 1e-2, 20000)     # 1 mm – 1 cm
medium = np.random.uniform(1e-2, 1e-1, 10000)    # 1 cm – 10 cm
large = np.random.uniform(0.1, 1, 3000)          # 10 cm – 1 m
satellite = np.random.uniform(1, 10, 500)        # 1 m – 10 m

sizes = np.concatenate([small, medium, large, satellite])


# ==============================
# SIZE → RCS CONVERSION
# ==============================
sigma_debris = []

for d in sizes:
    if d < wavelength / 10:
        # Rayleigh scattering region
        sigma = (np.pi**5 * d**6) / (wavelength**4)
    else:
        # Optical region
        sigma = np.pi * (d/2)**2
    
    sigma_debris.append(sigma)

sigma_debris = np.array(sigma_debris)


# ==============================
# PLOTTING
# ==============================
fig, ax1 = plt.subplots(figsize=(9,6))

# ---- N vs RCS ----
ax1.plot(sigma_vals, N_vals, linewidth=2, label='Required Antennas N')
ax1.set_xscale('log')
ax1.set_yscale('log')
ax1.set_xlabel("RCS σ (m²)")
ax1.set_ylabel("Number of Antennas (N)")
ax1.grid(True, which="both")

# ---- Histogram (same X-axis) ----
ax2 = ax1.twinx()
ax2.hist(sigma_debris, bins=120, alpha=0.3)
ax2.set_yscale('log')
ax2.set_ylabel("Object Count")

# ---- Optional Detection Limit ----
ax1.axhline(y=100, linestyle='--', label='Max Available Antennas')

# ---- Legend ----
ax1.legend()

plt.title("LEO Detection Capability: Antennas Required vs Object Distribution")
plt.tight_layout()
plt.show()
```
