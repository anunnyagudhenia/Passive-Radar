```python
#Arithmetic Operators
a=18
b=5
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a//b)
print(a%b)
print(a**b)
```

```python
#Assignment Operators
a=3
a+=1 # a=a+1
print(a)
a-=1 # a=a-1
print(a)
a*=2 #a=a*2
print(a)
a/=2 #a=a/2
print(a)
a%=2 #a=a%2
print(a)
a=5
a**=3 #a=a**3
print(a)
```

```python
#Comparison Operators
a=4
b=10
print(a==b)
print(a!=b)
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
```

```python
#Logical Operators
a=True
b=False
print(a and b)
print (a or b)
print(not b)
```

```python
#Bitwise Operator
print(111|89)
```
