Effective Isotropic Radiated Power

Effective Isotropic Radiated Power or Equivalent Isotropically Radiated Power (EIRP) is a measurement of radiated output power from an ideal isotropic antenna in a single direction.

An isotropic antenna is meant to distribute power equally in all directions – When we channel that power into a single direction and calculate the power it is known as EI

The power density arriving at the target depends on how much power the transmitter effectively radiates in that direction. If EIRP is low, very little energy reaches the target, resulting in weak reflected signals that may fall below the receiver noise floor.

Thus EIRP controls the first stage of the radar link budget: transmitter → propagation.RP

Radar Cross-section

The Radar Cross Section (RCS) is a measure of how much electromagnetic power a target scatters back toward the radar.

It is defined as the equivalent area that intercepts incident power and reradiates it isotropically toward the receiver.

RCS determines the strength of the reflected signal from the target. Large objects typically have large RCS values, while stealth objects are designed to have very small RCS values.

Thus RCS controls the target scattering stage of the radar signal chain.

```python
import numpy as np
import matplotlib.pyplot as plt

# Transmitter and Receiver positions (foci of ellipse)
Tx = np.array([-5.0, 0.0])
Rx = np.array([5.0, 0.0])

# Create grid in horizontal plane
x = np.linspace(-20, 20, 600)
y = np.linspace(-20, 20, 600)
X, Y = np.meshgrid(x, y)

# Distance from each grid point to Tx and Rx
RT = np.sqrt((X - Tx[0])**2 + (Y - Tx[1])**2)
RR = np.sqrt((X - Rx[0])**2 + (Y - Rx[1])**2)

# Bistatic range sum
L = RT + RR

# Iso-range values
levels = [12, 14, 16, 18, 20]

plt.figure()

# Draw iso-range contours
plt.contour(X, Y, L, levels=levels)

# Mark transmitter and receiver
plt.scatter(Tx[0], Tx[1])
plt.scatter(Rx[0], Rx[1])

plt.text(Tx[0], Tx[1], " Tx")
plt.text(Rx[0], Rx[1], " Rx")

plt.xlabel("x position")
plt.ylabel("y position")
plt.title("Bistatic Iso-Range Contours (RT + RR = constant)")
plt.axis("equal")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

lam = 1
d = lam/2

theta = np.linspace(-np.pi/2, np.pi/2, 2000)

# steering direction
theta0 = np.deg2rad(30)

def array_factor_eq5(N):

    psi = (2*np.pi*d/lam)*np.sin(theta)
    psi0 = (2*np.pi*d/lam)*np.sin(theta0)

    n = np.arange(N)

    # steering vector for observation angles
    a = np.exp(1j*np.outer(n,psi)) / np.sqrt(N)

    # complex weights steering toward theta0
    w = np.exp(-1j*n*psi0) / np.sqrt(N)

    AF = np.conjugate(w) @ a * np.sqrt(N)

    return np.abs(AF)**2


plt.figure()

for N in [2,8]:

    AF = array_factor_eq5(N)

    plt.plot(np.degrees(theta),
             10*np.log10(AF),
             label=f"N={N}")

plt.xlabel("Angle (degrees)")
plt.ylabel("Power (dB)")
plt.title("Array Factor using Equation (5) with Beam Steering")
plt.legend()
plt.grid()

plt.show()

import numpy as np
import matplotlib.pyplot as plt

lam = 1
d = lam/2

theta = np.linspace(-np.pi/2, np.pi/2, 2000)

def array_factor_eq6(N):

    psi = (2*np.pi*d/lam)*np.sin(theta)

    AF = (1/N)*(np.sin(N*psi/2)/np.sin(psi/2))**2

    AF[np.isnan(AF)] = N

    return AF


plt.figure()

for N in [2,8]:

    AF = array_factor_eq6(N)

    plt.plot(np.degrees(theta),10*np.log10(AF),label=f"N={N}")

plt.legend()
plt.xlabel("Angle (deg)")
plt.ylabel("Power (dB)")
plt.title("Equation (6): Closed-form Array Factor")

plt.grid()
plt.show()
```

```python
#e
import numpy as np
import matplotlib.pyplot as plt

# constants
kB = 1.38e-23
Tsys = 300

# receiver parameters
B = 200e3
tau = 1

# number of independent samples
n = int(B * tau)

# mean noise power
P = kB * Tsys * B

# simulate noise power samples (exponential distribution)
samples = np.random.exponential(P, n)

# running average
running_avg = np.cumsum(samples) / np.arange(1, n+1)

plt.figure()
plt.plot(running_avg)
plt.xlabel("Number of samples")
plt.ylabel("Estimated noise power")
plt.title("Radiometer Averaging of Noise Power")
plt.show()

# standard deviation reduction
std_single = np.std(samples)
std_mean = std_single / np.sqrt(n)

print("Number of samples:", n)
print("Processing gain:", np.sqrt(n))
print("Std of single sample:", std_single)
print("Std after averaging:", std_mean)
```
