
# Passive Radar – Theory and Derivations

This notebook presents the fundamental equations used in **passive radar systems**, with clear physical explanations and mathematical derivations.

---



# 1. Effective Isotropic Radiated Power (EIRP)

The **Effective Isotropic Radiated Power (EIRP)** represents the total power that an **ideal isotropic antenna** would need to radiate in order to produce the same power density as the real antenna.

### Definition

$$
EIRP = P_t G_t
$$

where

- $P_t$ = transmitted power  
- $G_t$ = transmitter antenna gain

### Physical Meaning

An isotropic antenna radiates power **equally in all directions**.  
Real antennas concentrate power in specific directions using **gain**, so the effective radiated power increases.

---



# 2. Power Density from an Isotropic Radiator

If power $P_t$ is radiated equally in all directions, it spreads over the surface of a sphere.

The surface area of a sphere is

$$
A = 4\pi R^2
$$

Therefore the **power density** at distance $R$ becomes

$$
S = \frac{P_t}{4\pi R^2}
$$

### Including Antenna Gain

For a directional antenna:

$$
S = \frac{P_t G_t}{4\pi R^2}
$$

Using the definition of EIRP:

$$
S = \frac{EIRP}{4\pi R^2}
$$

---



# 3. Antenna Effective Aperture

The **effective aperture** represents how much power an antenna can capture from an incoming electromagnetic wave.

The received power is

$$
P_r = S A_e
$$

where

- $S$ = incident power density
- $A_e$ = effective aperture of the antenna

### Relationship Between Gain and Aperture

From antenna theory:

$$
A_e = \frac{G \lambda^2}{4\pi}
$$

where

- $G$ = antenna gain  
- $\lambda$ = wavelength

This equation connects the **transmitting and receiving properties of antennas**.

---



# 4. Radar Cross Section (RCS)

The **Radar Cross Section (RCS)** describes how strongly a target reflects electromagnetic waves.

It is defined as

$$
\sigma = 4\pi R^2 \frac{P_{scattered}}{P_{incident}}
$$

where

- $\sigma$ = radar cross section
- $P_{incident}$ = power incident on the target
- $P_{scattered}$ = scattered power density

### Physical Meaning

A larger RCS means the target **reflects more energy back toward the radar**.

Typical values:

| Object | Approx RCS |
|------|------|
| Bird | 0.01 m² |
| Car | 5–10 m² |
| Aircraft | 10–100 m² |
| Satellite | 100–1000 m² |

---



# 5. Radar Equation (Bistatic / Passive Radar Form)

In passive radar, the transmitter and receiver are located at different positions.

The received power becomes

$$
P_r =
\frac{P_t G_t}{4\pi R_T^2}
\times
\sigma
\times
\frac{G_r \lambda^2}{(4\pi)^2 R_R^2}
$$

where

- $R_T$ = transmitter–target distance
- $R_R$ = target–receiver distance
- $G_r$ = receiver antenna gain

### Simplified Form

$$
P_r =
\frac{P_t G_t G_r \lambda^2 \sigma}
{(4\pi)^3 R_T^2 R_R^2}
$$

This is the **fundamental passive radar detection equation**.

---



# 6. Passive Radar Signal Chain

The passive radar signal flow is

```
Transmitter → Propagation → Target Scattering → Propagation → Receiver
```

### Steps

1. **Illuminator of opportunity** transmits radio waves.
2. Waves propagate to the **target**.
3. Target **scatters electromagnetic energy**.
4. Scattered signal propagates to the **receiver array**.
5. Receiver processes signal using **correlation and beamforming**.
