```python
import numpy as np
import matplotlib.pyplot as plt
```

Changing Frequency

```python
def freq():
    Pt = 10000  # Transmitter power in Watts
    Gt = 10     # Transmitter gain
    Gr = 6      # Receiver gain
    sigma = 100 # Radar Cross Section in m²
    Ld = 1      # Baseline distance factor
    Gcorr = 1   # Correlation gain
    Rt = 10000  # Distance between Transmitter and Object (m)
    Rr = 10000  # Distance between Receiver and Object (m)
    Ft = 1      # Propagation factor (transmitter to target)
    Fr = 1      # Propagation factor (target to receiver)

    X=[]
    Y=[]
    for freq in range(88, 101):
        wavelength = (3 * 10**8) / freq
        SNR = (Gt * sigma * (Ld **2) * Gcorr * Ft * (wavelength ** 2) *Gr)/((Rt**2) * (Rr **2) * ((4 * np.pi) **3) * (Ld**2))
        Y.append(SNR)
        X.append(freq)
    plt.plot(X,Y)
    plt.grid()
    plt.xlabel("Frequency")
    plt.ylabel("SNR")
freq()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Constants
Pt = 235000
Gt = 2511.89
lam = 1.453          # wavelength (206.5 MHz)
Rt = 300000
Rr = 300000
N0B = 1          

K = (Pt * Gt * lam**2) / ((4*np.pi)**3 * Rt**2 * Rr**2 * N0B)

# Yagi gains (linear)
G = [8, 10, 12.6, 15, 15.8, 20, 25.1]

sigma = np.linspace(1, 2500, 1000)

plt.figure()

for i, Gr in enumerate(G):
    SNR = K * sigma * Gr
    plt.plot(sigma, SNR, label=f'Ant {i+1}')

plt.xlabel("RCS (m²)")
plt.ylabel("SNR linear")
plt.title("Individual Antenna SNR vs RCS (7-element Yagi)")
plt.grid()
plt.legend()
plt.show()

for i, Gr in enumerate(G):
    SNR = K * sigma * Gr
    plt.plot(sigma, 10*np.log(SNR), label=f'Ant {i+1}')

plt.xlabel("RCS (m²)")
plt.ylabel("SNR dB")
plt.title("Individual Antenna SNR vs RCS (7-element Yagi)")
plt.grid()
plt.legend()
plt.show()
```

```python
SNR_fused = K * sigma * sum(G)
SNR_fused = K * sigma * sum(G)

plt.figure()
plt.plot(sigma, SNR_fused, 'k', linewidth=2)
plt.xlabel("RCS (m²)")
plt.ylabel("Fused SNR linear")
plt.title("Fused SNR vs RCS (7 Yagi Antennas)")
plt.grid()
plt.show()

plt.figure()
plt.plot(sigma, 10*np.log(SNR_fused), 'k', linewidth=2)
plt.xlabel("RCS (m²)")
plt.ylabel("Fused SNR linear")
plt.title("Fused SNR vs RCS (7 Yagi Antennas)")
plt.grid()
plt.show()
```

```python
plt.figure()

for M in [1, 3, 5, 7]:
    SNRm = K * sigma * sum(G[:M])
    plt.plot(sigma, SNRm, label=f'{M} antennas')

plt.xlabel("RCS (m²)")
plt.ylabel("Fused SNR linear")
plt.title("SNR vs RCS for Different Numbers of Yagi Antennas")
plt.grid()
plt.legend()
plt.show()

for M in [1, 3, 5, 7]:
    SNRm = K * sigma * sum(G[:M])
    plt.plot(sigma, 10*np.log(SNRm), label=f'{M} antennas')

plt.xlabel("RCS (m²)")
plt.ylabel("Fused SNR linear")
plt.title("SNR vs RCS for Different Numbers of Yagi Antennas")
plt.grid()
plt.legend()
plt.show()
```

```python
frequency = 206.5e6  #206.5MHz
c= 3e8
wavelength = c/frequency  #(m)
gr = np.array([8, 10, 12.6, 15, 15.8, 20, 25.1]) #Linear Yagi-Uda Gains
Pt = 235e3               #Watts
Gt = 2511.89              #Linear gain-34dB
Rt = 3e5
Rr = 3e5
k = 1.38e-23
T0 = 290
B = 5e6      #Bandwidth of Yagi Uda is between 2-5% of centre frequency
F = 3.16     #For practical receiver analysis, Fn is a factor greater than 1 (3-10 dB)- 5dB
sigma = np.linspace(1, 2500, 1000)
i=0
SNR_individual = []
for G in gr:
    SNR = (Pt * Gt * (wavelength**2) * G * sigma) / (((4*np.pi)**3) * k * T0 * B * F * (Rt**2) *(Rr **2))
    SNR_individual.append(SNR)
    plt.plot(sigma, 10*np.log(SNR), label=f'Ant{i+1}')
    i+=1

plt.xlabel("RCS (m²)")
plt.ylabel("SNR dB")
plt.title("Individual Antenna SNR vs RCS (7-element Yagi)")
plt.grid()
plt.legend()
plt.show()

   
```

```python
SNR_fused = np.sum(SNR_individual, axis=0)

plt.figure()
plt.plot(sigma, 10*np.log10(SNR_fused),label="Fused SNR (7 Yagi-Uda)")

plt.xlabel("RCS")
plt.ylabel("SNR (dB)")
plt.title("Fused SNR vs RCS")
plt.grid()
plt.legend()
plt.show()
```
