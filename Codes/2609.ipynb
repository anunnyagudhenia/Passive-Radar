```python
# Write a program to show the grades of a student
# if marks > 90 Grade=A
# if marks > 80 Grade=B
# if marks > 70 Grade=C
# if marks > 60 Grade=D
# if marks < 60 Fail
```

```python
marks = 81
if(marks > 90):
    print("A")
elif(marks > 80):
    print("B")
elif (marks > 70):
    print("C")
elif (marks > 60):
    print("D")
else:
    print("Fail")
```

```python
#While Loop
#Steps
#1. Check the condition
#2. If the condition is true, enter the loop. If the condition is false, don't enter the loop
#3. Repeat from 1.

i = 10 #initialize i
while (i>=1):
    print(i)
    i = i - 1
```

```python
# For Loop
#Steps
#1. Initialize
#2. Check the condition
#3. If the condition is true, enter the loop. If the condition is false, don't enter the loop
#4. Increment/ Decrement
#5. Repeat from 2

for i in range(11):
    print(i)
```
