Riemann Integral approximation is given by, 

$\int_{a}^{b} f(x) dx \approx \sum_{i=0}^{n-1} h f(x_i)$,

**2. Mid-point Rule** $\\$
If step size is $\Delta = (b-a)/n$, n is the number of sub-intervals, $\\$

$M_n = \sum_{i=1}^{n} f(m_i) \Delta x$.
