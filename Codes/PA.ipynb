## Effective Isotropic Radiated Power (EIRP)(Watt)

**Effective Isotropic Radiated Power (EIRP)**, also called **Equivalent Isotropically Radiated Power**, is the the measured radiated power in a single direction


If EIRP is low, very little energy reaches the target, resulting in weak reflected signals that may fall below the receiver noise floor.
A higher EIRP increases the illumination of the target and improves the overall detectability of objects in the radar system.

## Power Spectral Density (PSD)(Watt/Hz)

**Power Spectral Density (PSD)** describes how the **signal power is distributed across frequency**.

### Role of Power Spectral Density

Since noise is distributed over frequency, the **detectability of a signal depends on the signal power per unit frequency relative to the noise spectral density**.

If the signal power is spread over a wide bandwidth, the **power per unit frequency decreases**, making detection more difficult. Conversely, concentrating signal power into a narrower bandwidth improves the **signal-to-noise ratio (SNR)**.

Thus, **Power Spectral Density directly controls the signal-to-noise characteristics of the received signal** and plays an important role in determining whether a target can be detected.

## Antenna Effective Aperture($m^2$)

The **effective aperture** (or **effective area**) of an antenna represents the amount of incident electromagnetic power that the antenna can capture from an incoming electromagnetic waveruare meters } (m^2)
\]

### Role in Radar Detection

The effective aperture determines how much of the **incident power flux density** is converted into **received electrical power**.

Even if the incoming electromagnetic wave is strong, an antenna with a **small effective aperture** captures only a small portion of the available power. Larger antennas therefore collect more energy and improve the **sensitivity of the radar receiver**.

Thus, the effective aperture controls the **receiver intercerightarrow \text{Received Power}
\]

# Radar Cross Section (RCS)($m^2$)

The **Radar Cross Section (RCS)** is a measure of how much electromagnetic power a target scatters back toward the radar receiver.



### Role

The radar cross section determines the **strength of the reflected signal** received by the radar system.

- **Large objects** (aircraft, ships, satellites) generally have **large RCS values**, making them easier to detect.
- **Stealth objects** are designed to **minimize RCS**, reducing the amount of scattered energy directed back toward the radar.

The **bistatic angle** $\beta$ is defined as the angle at the target between the directions to the transmitter and the receiver.

The Doppler shift occurs because the **total propagation path length changes as the target moves**.

If the target moves with velocity \(v\), the rate of change of the bistatic path length is

$$
\frac{dL}{dt} = \frac{dR_T}{dt} + \frac{dR_R}{dt}

### Doppler Frequency Shift

The Doppler frequency shift is related to the rate of change of the propagation path length:

$$
f_D = \frac{1}{\lambda}\frac{dL}{dt}
$$

  

### Vector Form of Range Rate

Using vector geometry, the rate of change of each range component is

$$
\frac{dR_T}{dt} = \hat{R}_T \cdot v
$$

$$
\frac{dR_R}{dt} = \hat{R}_R \cdot v
$$

  



Substituting these into the Doppler equation gives

$$
f_D = \frac{1}{\lambda} (\hat{R}_T + \hat{R}_R) \cdot v
$$
### (b) 

Forward scatter occurs when 

$$
\beta = 180^\circ
$$

This means the **transmitter and receiver are located on opposite sides of the target**.

In this configuration, the vector

$$
\hat{R}_T + \hat{R}_R
$$

is perpendicular to the **transmitter–receiver baseline**.

### Doppler Shift in Forward Scatter

The Doppler shift simplifies to

$$
f_D = \frac{2}{\lambda} v_{\perp}
$$

where

- \(f_D\) = Doppler frequency shift  
- \(\lambda\) = wavelength  
- \(v_{\perp}\) = component of the target velocity **perpendicular to the transmitter–receiver baseline**

### Physical Interpretation

In **forward scatter geometry**:

- The target behaves like an **obstacle producing diffraction** of the incident wave.
- The scattered electromagnetic field can become **very strong in the forward direction**.
- This leads to **forward-scatter radar cross section enhancement**.

Because of this enhancement, **passive radar systems can detect very small or low-RCS targets**, including objects designed to be difficult to detect by conventional monostatic radar.
### (c) Bistatic Range Resolution

**Range resolution** determines the ability of a radar system to distinguish two targets separated in distance.

For a **bistatic radar**, the range resolution is

$$
\Delta R_b = \frac{c}{2B\cos(\beta/2)}
$$

where

- \(c\) = speed of light  
- \(B\) = signal bandwidth  
- \(\beta\) = bistatic angle  

### Effect of Bistatic Angle

As the bistatic angle approaches

$$
\beta \rightarrow 180^\circ
$$

then

$$
\cos(\beta/2) \rightarrow 0
$$

which causes the bistatic range resolution to become

$$
\Delta R_b \rightarrow \infty
$$

### Geometric Interpretation

The bistatic range measurement corresponds to **elliptical contours** defined by

$$
R_T + R_R = \text{constant}
$$

where

- \(R_T\) = transmitter-to-target distance  
- \(R_R\) = target-to-receiver distance  

Near **forward-scatter geometry**, these ellipses become **highly elongated**. As a result, many spatial points correspond to nearly the **same bistatic range measurement**.

### Implication

Because of this geometric effect, the radar cannot effectively **distinguish targets separated along the elongated direction**, leading to **poor range resolution near forward scatter**.

### (d) Grating Lobes in an Antenna Array

**Grating lobes** occur when multiple directions satisfy the condition for **constructive interference** in an antenna array.

Constructive interference occurs when the phase difference between adjacent elements is an integer multiple of \(2$\pi$:

$$
\psi = 2\pi m
$$

where

- \(m\) = integer (0, ±1, ±2, ...)

### Substituting the Phase Expression

The phase difference between elements is

$$
\psi = \frac{2\pi d}{\lambda}\sin\theta
$$

Substituting into the constructive interference condition:

$$
\frac{2\pi d}{\lambda}\sin\theta = 2\pi m
$$

Simplifying gives

$$
\sin\theta = \frac{m\lambda}{d}
$$

### Condition for Avoiding Grating Lobes

For physically valid directions,

$$
-1 \le \sin\theta \le 1
$$

To ensure that **only the main lobe exists within the visible region**, we require that no additional integer \(m\) produces a valid solution.

This leads to the spacing constraint

$$
d \le \frac{\lambda}{2}
$$

```python
import numpy as np
import matplotlib.pyplot as plt

lam = 1
d = lam/2
k = 2*np.pi/lam

theta = np.linspace(-np.pi/2, np.pi/2, 2000)


# ---------------------------------------------------
# Equation (5) : AF = w^H a(theta) / sqrt(N)
# ---------------------------------------------------

def array_factor_eq5(N):

    n = np.arange(N)

    w = np.ones(N)/np.sqrt(N)          # uniform weights

    AF = np.zeros_like(theta,dtype=complex)

    for i,t in enumerate(theta):

        a = np.exp(1j*k*d*n*np.sin(t))  # steering vector

        AF[i] = np.conjugate(w) @ a / np.sqrt(N)

    return np.abs(AF)**2


# ---------------------------------------------------
# Equation (6) : closed-form array factor
# ---------------------------------------------------

def array_factor_eq6(N):

    psi = (2*np.pi*d/lam)*np.sin(theta)

    AF = (1/N)*(np.sin(N*psi/2)/np.sin(psi/2))**2

    AF[np.isnan(AF)] = N

    return AF


# ---------------------------------------------------
# Plot Linear Arrays
# ---------------------------------------------------

plt.figure()

for N in [2,8]:

    AF = array_factor_eq6(N)

    plt.plot(np.degrees(theta),10*np.log10(AF),label=f"N={N}")

plt.xlabel("Angle (deg)")
plt.ylabel("Power (dB)")
plt.title("ULA Beam Pattern (d = λ/2)")
plt.legend()
plt.grid()

plt.show()


# ---------------------------------------------------
# Equation (7) : theoretical 3-dB beamwidth
# ---------------------------------------------------

for N in [2,8]:

    bw_rad = 0.886/N
    bw_deg = 50.8/N

    print(f"N={N}")
    print(f"3-dB Beamwidth ≈ {bw_rad:.3f} rad")
    print(f"3-dB Beamwidth ≈ {bw_deg:.2f} deg\n")


# ---------------------------------------------------
# Hexagonal Array (7 elements)
# ---------------------------------------------------

phi = np.linspace(0,2*np.pi,360)

positions = np.array([
[0,0],
[d,0],
[d/2, np.sqrt(3)*d/2],
[-d/2, np.sqrt(3)*d/2],
[-d,0],
[-d/2,-np.sqrt(3)*d/2],
[d/2,-np.sqrt(3)*d/2]
])

AF = np.zeros_like(phi,dtype=complex)

for x,y in positions:

    AF += np.exp(1j*k*(x*np.cos(phi)+y*np.sin(phi)))

AF = np.abs(AF)/len(positions)

power = AF**2

plt.figure()

plt.polar(phi,power)

plt.title("Hexagonal Array Beam Pattern (7 elements)")

plt.show()
```

### (c) Noise Spectral Density

The **noise spectral density** is defined as

$$
N_0 = \frac{P_N}{B}
$$

where

- $N_0$ = noise spectral density  
- $P_N$ = total noise power  
- $B$ = bandwidth  

### Substituting Thermal Noise Power

From the Johnson–Nyquist theorem,

$$
P_N = k_B T B
$$

Substituting into the spectral density expression gives

$$
N_0 = k_B T
$$

### Two-Sided Noise Spectrum

In signal processing, the noise spectrum is usually considered **two-sided**, meaning noise exists at both positive and negative frequencies.

Therefore the spectral density is split equally between the two halves:

$$
\frac{N_0}{2} = \frac{k_B T}{2}
$$

For a receiver with **system temperature \(T_{sys}\)**,

$$
\frac{N_0}{2} = \frac{k_B T_{sys}}{2}
$$

### Noise Power from Spectral Density

The total noise power is obtained by integrating the spectral density over the receiver bandwidth:

$$
P_N = \int_{-B/2}^{B/2} N_0 \, df
$$

Since thermal noise is **white**, the spectral density is constant:

$$
N_0 = k_B T_{sys}
$$

Substituting into the integral:

$$
P_N = k_B T_{sys} \int_{-B/2}^{B/2} df
$$

Evaluating the integral gives

$$
P_N = k_B T_{sys} B
$$

Thus the **total noise power in a receiver** is

$$
P_N = k_B T_{sys} B
$$

The beamformer uses the **steering vector**

which aligns the phases of signals arriving from the target direction.

Each antenna element receives the same signal with a phase shift.



After beamforming, the signals from all elements are summed:

$$

Because the steering vector aligns phases,

$$
s_{sum}(t) = N\,s(t)
$$

Thus the **signal amplitude increases by a factor \(N\)**.

$$


### Why Signal Adds Coherently

The steering vector

$$
a(\theta_0)
$$

compensates for phase delays across the array, so signals arriving from the target direction are **in phase when summed**.

Thus before normalization:

- Amplitude $ \propto N $
- Power $ \propto N^2 $



### Noise Behavior

Receiver noise at each element is **independent and uncorrelated**.

Given - Random phases mean the noise amplitudes combine as

$$
\sqrt{N}
$$

Thus noise power becomes

$$
P_{noise} \propto N
$$


### Resulting SNR Improvement

- Signal power grows as \(N^2\) (before normalization)
- Noise power grows as \(N\)

Therefore the **signal-to-noise ratio improves by**

$$
\text{SNR gain} = N
$$

This improvement is called the **array gain**.

Passive radar uses **cross-correlation processing** between two channels:

### (e) Practical Design Parameters in Passive Radar

In passive radar systems, the operator cannot control several key parameters, including:

- $P_{\text{EIRP}}$ — transmit power of the broadcast transmitter  
- $R_T$ — distance between the transmitter and the target  
- $\lambda$ — transmitter operating wavelength (frequency)

Because these parameters are determined by the external transmitter, system performance must be optimized through receiver-side design choices.

### Controllable Design Parameters

| Parameter | Optimization Strategy | Practical Constraint |
|----------|----------------------|---------------------|
| \(N\) (array size) | Increase number of antenna elements to improve array gain | Higher hardware cost and calibration complexity |
| \(\tau\) (integration time) | Increase integration time to improve detection sensitivity | Limited by target motion |
| \(T_{sys}\) (receiver noise temperature) | Reduce receiver noise to improve SNR | Limited by LNA performance and practical cooling constraints |

### Array Directivity

Directivity measures how concentrated the radiation pattern is.

For an array with uniform weights, signals from the elements add coherently in the main beam.
When N identical elements are combined, signal amplitude increases by N and signal power increases by N

$$
G_{array} \approx N
$$

If each element has gain $𝐺_𝑟$, the total array directivity becomes

$$
G_{array} \approx N G_r
$$

For isotropic antenna elements,

$$
G_r = 1
$$

Thus the array directivity becomes

$$
D_{array} \approx N
$$

The **effective aperture of a single antenna element** is

$$
A_e = \frac{G_r \lambda^2}{4\pi}

For an array with \(N\) elements, the **total effective aperture** becomes

A_{tot} = N A_e

Substituting the expression for \(A_e\):

$$
A_{tot} = N \frac{G_r \lambda^2}{4\pi}

Rearranging gives

$$
N G_r \lambda^2 = 4\pi A_{tot}

From the bistatic radar sensitivity expression,

$$

\sigma_{b,min} =

In coherent beamforming, the array provides an additional gain proportional to

Thus the minimum detectable bistatic radar cross section scales as
