```python
import skrf as rf
#rf.Network(...) loads Touchstone S1P file into an RF network object.
ntwk = rf.Network('anuS1P1.s1p')
ntwk
```

```python
freq_mhz = ntwk.f / 1e6

# S11 in dB
s11_db = 20 * np.log10(np.abs(ntwk.s[:, 0, 0]))

# Plot S11
idx_min = np.argmin(s11_db)

dip_freq = freq_mhz[idx_min]
dip_s11 = s11_db[idx_min]

print(f"Dip frequency: {dip_freq:.2f} MHz")
print(f"Minimum S11: {dip_s11:.2f} dB")
plt.figure()
plt.plot(freq_mhz, s11_db)
plt.axvline(206.5, linestyle='--',label='206.5mhz')  # target frequency
plt.xlabel('Frequency (MHz)')
plt.ylabel('S11 (dB)')
plt.title('S11 vs Frequency')
plt.legend()
plt.grid()
plt.show()
```

```python
# Reflection coefficient
gamma = np.abs(ntwk.s[:, 0, 0])

# VSWR
vswr = (1 + gamma) / (1 - gamma)

plt.figure()
plt.plot(freq_mhz, vswr)
plt.axvline(206.5, linestyle='--')
plt.xlabel('Frequency (MHz)')
plt.ylabel('VSWR')
plt.title('VSWR vs Frequency')
plt.ylim(1, 5)
plt.grid()
plt.show()
```
